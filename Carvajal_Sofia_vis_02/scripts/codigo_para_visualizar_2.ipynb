```python
!pip -q install altair vl-convert-python Pillow

import io
import pandas as pd
import altair as alt
from PIL import Image

# Datos integrados directamente para evitar errores de archivo no encontrado
csv_data = """club,minutos_sub21,jugadores_sub21,posicion_predominante,min_defensas,min_volantes,min_delanteros,min_arqueros,cumplio_regla
Universidad Catolica,2185,4,Delantero,180,780,1225,0,Si
Huachipato,3150,5,Volante,1080,1350,720,0,Si
O Higgins,3370,4,Volante,890,1940,540,0,Si
Curico Unido,2990,3,Lateral,1680,0,1310,0,Si
Union Espanola,2420,4,Volante,420,1620,380,0,Si
Colo-Colo,3640,5,Lateral,1890,1070,680,0,Si
Universidad de Chile,2520,5,Delantero,360,950,1120,90,Si
Union La Calera,1540,3,Lateral,810,490,240,0,Si
Everton,1860,3,Lateral,1240,340,280,0,Si
Palestino,1490,3,Volante,310,870,310,0,Si
Audax Italiano,1410,2,Volante,210,920,280,0,Si
Coquimbo Unido,1390,3,Delantero,150,410,830,0,Si
Cobresal,1380,2,Volante,240,860,280,0,Si
Deportes Antofagasta,1420,3,Delantero,280,390,750,0,Si
Deportes Iquique,1360,2,Volante,120,760,480,0,Si
Universidad de Concepcion,1370,3,Volante,190,820,360,0,Si"""

df = pd.read_csv(io.StringIO(csv_data))

bars = (
    alt.Chart(df)
    .mark_bar(cornerRadiusTopRight=3, cornerRadiusBottomRight=3)
    .encode(
        y=alt.Y("club:N", sort="-x", title="Club Primera División"),
        x=alt.X("minutos_sub21:Q", title="Minutos totales Sub-21 otorgados en 2019"),
        color=alt.Color(
            "posicion_predominante:N",
            title="Posición más utilizada",
            scale=alt.Scale(
                domain=["Volante", "Lateral", "Delantero"],
                range=["#0072B2", "#E69F00", "#009E73"]
            )
        ),
        tooltip=[
            alt.Tooltip("club:N", title="Club"),
            alt.Tooltip("minutos_sub21:Q", title="Minutos Sub-21", format=","),
            alt.Tooltip("jugadores_sub21:Q", title="Canteranos usados"),
            alt.Tooltip("posicion_predominante:N", title="Posición predominante"),
            alt.Tooltip("min_volantes:Q", title="Min. Volantes", format=","),
            alt.Tooltip("min_defensas:Q", title="Min. Defensas/Laterales", format=","),
            alt.Tooltip("min_delanteros:Q", title="Min. Delanteros", format=","),
        ]
    )
)

regla_min = (
    alt.Chart(pd.DataFrame({"x": [1350]}))
    .mark_rule(strokeDash=[4, 4], color="#D55E00", strokeWidth=2)
    .encode(x="x:Q")
)

texto_regla = (
    alt.Chart(pd.DataFrame({"x": [1350], "y": ["Universidad de Concepcion"]}))
    .mark_text(text="Mínimo exigido ANFP (1.350 min)", align="left", dx=5, dy=15, color="#D55E00", fontSize=11, fontWeight="bold")
    .encode(x="x:Q", y="y:N")
)

chart = (
    (bars + regla_min + texto_regla)
    .properties(
        width=680,
        height=450,
        title=alt.TitleParams(
            "Minutaje juvenil por club: cumplimiento administrativo vs. convicción formativa (2019)",
            subtitle=["Minutos totales otorgados a jugadores Sub-21 por los 16 clubes de Primera División.",
                      "La línea naranja discontinua marca la cuota obligatoria ANFP de 1.350 minutos."],
            anchor="start"
        )
    )
    .configure_axis(labelFontSize=11, titleFontSize=12)
    .configure_legend(labelFontSize=11, titleFontSize=11)
)

# Guardar los archivos de salida
chart.save("vis_02.html")
chart.save("vis_02.png", scale_factor=2)
Image.open("vis_02.png").convert("RGB").save("vis_02.jpg", quality=95)

print("¡Listo! vis_02.html y vis_02.jpg generados con éxito.")
chart
```
